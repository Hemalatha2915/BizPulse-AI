```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop",
                "Mouse", "Monitor", "Keyboard", "Monitor"],
    "Quantity": [2, 10, 5, 1, 15, 3, 8, 2],
    "Price": [50000, 500, 1500, 50000, 500, 12000, 1500, 12000],
    "Cost": [42000, 250, 900, 42000, 250, 9000, 900, 9000]
}

df = pd.DataFrame(data)
df["Revenue"] = df["Quantity"] * df["Price"]
df["Profit"] = df["Quantity"] * (df["Price"] - df["Cost"])

print("Sales data created successfully!")
display(df)
```

```python
product_profit = df.groupby("Product")["Profit"].sum()
product_profit = product_profit.sort_values(ascending=False)

print("Product-wise Total Profit:")
print(product_profit)

print("\nMost Profitable Product:", product_profit.idxmax())
print("Highest Total Profit:", product_profit.max())
```

```python
import matplotlib.pyplot as plt

product_profit.plot(kind="bar")

plt.title("BizPulse AI - Product Profit Analysis")
plt.xlabel("Product")
plt.ylabel("Total Profit (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

product_analysis = df.groupby("Product")[["Revenue", "Profit"]].sum()

product_analysis.plot(kind="bar", figsize=(10, 5))

plt.title("BizPulse AI - Revenue vs Profit")
plt.xlabel("Product")
plt.ylabel("Amount (₹)")
plt.xticks(rotation=0)
plt.legend(["Revenue", "Profit"])
plt.tight_layout()
plt.show()
```

```python
product_analysis["Profit_Margin_%"] = (
    product_analysis["Profit"] /
    product_analysis["Revenue"]
) * 100

print("BizPulse AI - Profit Margin Analysis")

display(product_analysis.round(2))
```

```python
total_revenue = df["Revenue"].sum()
total_profit = df["Profit"].sum()
overall_margin = (total_profit / total_revenue) * 100

print("===== BizPulse AI | Business Summary =====")
print(f"Total Revenue: ₹{total_revenue:,.0f}")
print(f"Total Profit: ₹{total_profit:,.0f}")
print(f"Overall Profit Margin: {overall_margin:.2f}%")
print(f"Products Analysed: {df['Product'].nunique()}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))
fig.suptitle("BizPulse AI | Business Dashboard",
             fontsize=16, fontweight="bold")

kpis = [
    ("Total Revenue", f"Rs. {total_revenue:,.0f}"),
    ("Total Profit", f"Rs. {total_profit:,.0f}"),
    ("Profit Margin", f"{overall_margin:.2f}%")
]

for ax, (title, value) in zip(axes, kpis):
    ax.set_facecolor("#171D32")
    ax.text(0.5, 0.62, title, ha="center",
            fontsize=12, color="#C4B5FD")
    ax.text(0.5, 0.38, value, ha="center",
            fontsize=17, fontweight="bold", color="white")
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(False)

fig.patch.set_facecolor("#0B1020")
plt.tight_layout()
plt.show()
```
